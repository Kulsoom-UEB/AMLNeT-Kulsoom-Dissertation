# Notebook 05: Model Development and Hyperparameter Tuning

**Project:** Explainable Machine Learning for Suspicious Transaction Risk Detection in Anti-Money Laundering

**CRISP-DM stage:** Modelling

**Dataset:** AMLNet (Huda et al., 2025), Zenodo [10.5281/zenodo.16736515](https://doi.org/10.5281/zenodo.16736515)

### Purpose

Three algorithms (Logistic Regression, Random Forest, XGBoost) trained under two imbalance strategies, with RandomizedSearchCV tuning scored by PR-AUC on the training set only.

### Key points

* Class weighting: scaling + one-hot **before** the model
* SMOTENC: scaling → SMOTENC → one-hot **after** resampling
* SMOTENC (not plain SMOTE) because `type` and `category` are categorical
* LightGBM removed from project scope
* Six tuned model/strategy combinations plus a majority-class benchmark
